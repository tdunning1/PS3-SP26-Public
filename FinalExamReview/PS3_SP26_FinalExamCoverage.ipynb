## 1. Coverage and format of the final exam


### Coverage
The coverage of the final exam is cumulative. This means that all the course material is fair game, unless I have explicitly said something would not be covered (such as the `weighted.mean()` function in the Notebook Lecture for Week 14).

The best preparation for the exam is to review (1) all the in-class activities and problem sets, (2) lecture slides and Notebook Lectures, (3) the required readings (and the optional readings may help you), (4) your section materials, and (5) all other assignments and aspects of the course.

#### Format
The format will be similar to the midterm:
- You will complete the exam in a Jupyter Notebook and will upload a zip file to Gradescope, just like you have done for in-class assignments and problem sets and as you did on the midterm. The exam will also be autograded, similar to other assignments in the class. 
- There will be a mix of multiple-choice conceptual questions and coding questions. There will be a slightly greater proportion of coding questions on the final than on the midterm---but both kinds of questions will be on the exam.
- The exam will ask you to load a dataset or datasets **that we have already used in the course,** conduct analysis of the data, and answer conceptual questions about your analysis. **We will not use any new data that we have not already used in an in-class activity or problem set.**
- Where we have learned multiple techniques for accomplishing the same task -- such as the "old old" way of calculating differences of means, using `difference_in_means()`, or using bivariate regression -- the exam will tell you which one to use in a given question.
- One or two of the data analysis questions (but no more than that) will let you use Otter cells to check you answer.

For multiple-choice questions, as on the midterm, 
- Remember to use a process of elimination if needed to find the answer!
- Also, pay attention to whether the question is asking for the option that IS correct or is NOT correct!

As on the midterm, you might need to think a bit about some of the questions---but if you have studied the material, I feel confident you can answer them!!!

#### Calculating the course grade

Per the syllabus, your final grades will be based on the following:

- In-Class Individual Assignments: 20%
    - Your lowest individual assignment grade will be dropped.
    - The first assignment will not count towards your final grade.
- In-Class Group Assignments (0%)
    - These do not count towards the official percentage point total, but completion of the assignments will be recorded. However, course grades will be lowered for members of groups who fail to submit more than two assignments on time.
- Problem Sets: 15%
- Final Project: 10%
- Midterm Exam: 20%
- Final Exam: 30%
- Section Attendance: 5%

For **calculating your score on the individual in-class assignments,** we do the following:
- Convert each score to a proportion/percentage.  For example, 5/10 is 0.5, 10/15 is 0.666, etc.
- Replace the score with "NA" if you had a valid excuse (sports, medical, etc) for missing the assignment; otherwise, if the score is missing it is a 0.
- We drop the first assignment (since it doesn't count towards the final grade, per the syllabus) and gave everyone who completed the second assignment a score of 1 (i.e. 100%). 
- Drop your lowest score (the assignment on which you had the lowest proportion/percentage).
- Take the average of the remaining scores (the average of the proportion/percentages).

Note that we will apply extra credit points after doing this calculation and applying a curve to it. Thus, as I said, your grade is helped by your extra credit points, even if everybody does the extra credit assignments we offered.

The course grade will be calculated by taking a weighted average of each component of the course, using the weights/percentages noted above. 

After we have calculated this weighted average score for each student, we will apply the curve to the distribution of scores to come up with the final letter grade you would get without any extra credit.

Then, we will recalculate your score with the extra credit points included to get your final grade.
