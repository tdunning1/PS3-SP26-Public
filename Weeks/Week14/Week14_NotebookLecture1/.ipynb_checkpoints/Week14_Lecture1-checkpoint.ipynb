### Why generalize?

We have talked a lot in this course about how to draw causal inferences successfully.  For example, we have focused on how experiments can help us make inferences about cause and effect.

But when we conduct experiments, we learn about the effect of the *particular treatment* studied, conducted *among the people* the experiment was conducted among, *at the time/in the context* the experiment was conducted. 

However, we're also often interested in *generalizing* the experimental results to places where these dimensions of the experiment are different.  For example, we might want to know what the effects would be when there are:

- **Different treatments**: Remember that in Problem Set 2, you looked at the effect of a "gentle" social pressure mailer (that was gentler than the example we discussed in class in Week 5). But what if we made it even "gentler" by not telling people how often their neighbors had voted?
- **Different subjects/units/people**: Remember we discussed the experiment in Utah in which Republican precinct chairs used different strategies to recruit women delegates to the state party convention.  But would the effects of these treatments be similar for Utah Democratic precinct chairs? How about precinct chairs in other states?
- **Different times/context**: The social pressure experiment we analyzed in Week 5 was conducted in an off-year primary election in Michigan. Would the effects be different in a Presidential general election (e.g., Biden versus Trump in 2020)?

The ability of a study to tell us about the causal effect of a particular treatment among a particular set of people in a given time or context is sometimes referred to as the study's "**internal validity**."  Generalizing to other, related treatments or new people or contexts relates to what is sometimes called "**external validity**".

In general, "internal validity"---that is, valid causal attribution within a given study---can be challenging. We've talked in this course about a lot of reasons internal validity might fail, such as omitted variables bias in observational studies.  

But assessing "external validity" or generalizability with evidence is sometimes even harder. 

However, this does not mean that it is impossible. In this lecture, I will talk about some of the techniques social science researchers use to assess generalizability. 

So in this lecture, we are going to learn about 
1. why effects might not generalize perfectly;
2. one common approach for studying generalizability (called "meta-analysis");
3. some of the difficulties of trying to understand *why* effects might vary across different treatments, people, or contexts;
4. a statistical danger that arises when we compare effects across a lot of different treatments, types of subjects, or contexts (called $p$-hacking); and
5. The value of *replication* of experiments as well as of introducing *controlled variation* to get at the *why* question.

By the 2020 election, social pressure mail had become a mainstay of political campaigns, and so voters in swing states were recieving a lot of social pressure mail. Wisconsin was an important swing state in 2020, so voters there likely received a lot of social pressure mail.

In both places, the experiments are completely valid. In both places, they would answer the same question: what is the effect of sending voters a social pressure mailer on voter turnout?

But they are answering those questions in different contexts. And in Wyoming, the effect of sending voters a social pressure mailer might be much bigger because voters otherwise wouldn't get any social pressure mailers. But in Wisconsin, the effect might be smaller because people there probably already received a lot of social pressure mailers.

At least, that is my guess of what would happen. But, **there is no magic way to know whether the effect of a treatment would be the same if it was done differently, among other people, or in another time/context. We just have to do our best to make an educated guess -- or gather more data to understand how a treatment's effects vary.**

### A common approach for understanding generalizability: meta-analysis

Is generalizing hopeless? 

Instead of just throwing up our hands and saying we don't know what one experiment says about what causal effects would be of another treatment, in another context, or among a different population, a common approach is to look across the results of *many* experiments and look to see how the results vary.

This is called *meta-analysis* because it's an analysis of analyses.

For some questions, we can't do this, because there's only been one experiment -- all we can do is make an educated guess.

But for other questions that are studied more often, we can compile evidence from across many contexts and understand if and how effects vary.

Today, we'll take a look at a dataset from a recent paper that collects all of the public experiments that had been done on persuasion in American political campaigns.

### Campaign persuasion dataset

```r
# RUN THIS CELL

options(warn = -1) # suppress warnings

library(estimatr) # load the estimatr package
library(ggplot2) # load the ggplot2 package

data <- read.csv("ps3_KB_campaign.csv") # read in the data
head(data) # show the first six rows of the data
```

In this dataset, every row is an experiment. The authors of this study (David Broockman and Josh Kalla) collected information from previous experiments and collected it in this dataset for meta-analysis.

Here is a quick rundown of what each column means:

- `Experiment`: The particular experiment (the reference for the study with the experiment corresponding to this row of the dataset)
- `SE`: standard error of the estimate of the effect size in the corresponding experiment
- `effect`: estimate of the effect size in the corresponding experiments
- `precision`: precision; this is how precise the experiment is. defined as $\frac{1}{\text{Standard Error}^2}$ (these are weights to be used in our analysis)
- `General`: a 0-1 (binary) variable indicating whether the experiment was done in a general election (1 = yes, 0 = no)
- `Primary`: a 0-1 (binary) variable indicating whether the experiment was done in a primary election (1 = yes, 0 = no). these elections are typically quieter
- `Treatment.within.2months.of.election`: binary variable incating whether the experiment took place within 2 months of election day (1 = yes, was within 2 months; 0 = no, was more than 2 months before election day)

### Meta-analysis using `weighted.mean()`

One way we could analyze this data is by just looking at the average effect -- i.e., the average of the estimated effects across all of these experiment.

```r
mean(data$effect)
```

But some of these experiments are much more precise than others:

```r
qplot(data$SE, binwidth=02)
```

This could be, for example, because some studies are much larger than others (have a bigger N), so they are more precise.

When we conduct meta-analyses, we may want to *weight* our estimates by how precise they are.  

Now, we're going to talk more now about what *weighting* means and how to do it.  **You won't need to use the weighting code yourselves on the final exam,** since we won't be practicing this with an in-class activity. But **you are responsible for understanding the idea** -- and for knowing what is happening when we take a *weighted average*, i.e., a weighted mean.

So why might we want to *weight* our estimates by how precise they are? Consider this example we discussed earlier in the semester:

If we wanted to combine the data from these two datasets to ask "what is the average effect of letters", we wouldn't want to just average 8% and 5%. We have a lot more information about the effects of letters from Experiment 2, so we'd want to put a lot more weight on experiment 2.

It turns out that the amount of information in an experiment -- how *precise* it is -- is defined as $\frac{1}{\text{Standard Error}^2}$. The basic message: the smaller the standard error, the more precise the experiment is.

```r
example.data <- data.frame(experiment = c(1,2), estimate = c(8, 5), SE = c(30, 1))
example.data
```

Adding in a precision column:

```r
example.data$precision <- 1/example.data$SE^2
example.data
```

Unweighted, the results would look like this:

```r
mean(example.data$estimate)
```

To weight our average by this, we can use the `weighted.mean()` function, like this:

```r
weighted.mean(example.data$estimate, example.data$precision)
```

Notice that the weighted average is much closer to the more precise estimate---i.e., 5---than it is to the less precise estimate---i.e., 8.

Here `weighted.mean()` takes average of the `estimate` variable in `example.data` and uses the `precision` a weight.  

In the background, to get the weighted average you (1) take the product of each element in the average and its weight (that is, multipy them together); (2) add all the products together; and (3) divide what you added by the sum of the weights.  

So we could get the same answer by doing this calculation directly:

```r
ave_weighted <- (8*(0.001111111) + 5*(1.000000000))/sum(example.data$precision)
ave_weighted                                                       
```

The answers are very slightly different because we rounded the elements in the sum.  But the answers are the same to the eighth digit:

```r
round(ave_weighted,8)
```

Let's apply `weighted.mean()` to the main dataset we loaded, i.e. the data set of experiments:

```r
weighted.mean(data$effect, data$precision)
```

#### Using `weighted.mean()` to examine how effects vary

To examine how effects vary and generalize, we can also run `weighted.mean()` function in different subsets. 

For example, in this example, let's look at effects in Primary elections and General elections.

In primary elections:

```r
primaries <- subset(data, Primary == 1)
weighted.mean(primaries$effect, primaries$precision)
```

Now, in general elections:

```r
general.elections <- subset(data, General == 1)
weighted.mean(general.elections$effect, general.elections$precision)
```

Interesting! Looks like the effects of campaign persuasion are much larger in primaries than in general elections.

This is a good example of a lesson: effects in one context doesn't necessarily generalize to others!

#### Understanding _why_ effects vary

It's not necessarily straightforward to understand _why_ effects vary across people, contexts, etc. from meta-analyses.

For example, why are the effects of campaign persuasion larger in primaries than in general elections?

This could be for a lot of reasons: in general elections, voters might remain loyal to their parties; but in primaries, candidate's party doesn't appear on the ballot, so more voters might be "up for grabs." Alternatively, in general elections tend to have more campaign spending, so the effect of additional spending might be lower than in primaries, which tend to be quieter.

One of the things meta-analyses do is prompt further theories and hypotheses to understand what creates the patterns we find in them.

For example, based on the evidence from the meta-analyses above, we might theorize that, as I said above, when candidates' parties don't appear on a ballot, it's easier to persuade voters. To test this hypothesis, one could imagine an experiment showing voters a sample ballot and randomly assigning both a) whether or not candidate party labels are present and b) information about the candidates. We could then look to see if the effect of the information is bigger when candidate party labels are shown versus when they are not.

### Caution: $p$-hacking

It is important to note that there is a statistical danger with comparing effects across different sub-groups.  

In general, there could be lots of different types of sub-groups we could compare.  

For example, they could be different types of individuals in a single experiment.  Thus, we might compare treatment effects for Democrats and for Republicans in an experiment in order to assess whether there are difference effects for these two types of subjects. (These differences are often called *heterogeneous treatment effects*). We've done these types of comparisons earlier in this class.

Or, we could compare results in different types of experiments---for instance, those run in primaries vs. those run in general elections, like we did here. 

This danger is often called ***p*-hacking**.

The idea of $p$-hacking is that, if you test lots and lots of hypotheses that are not true (i.e., where the null hypothesis is true), you're likely to get at least one statistically significant result just by chance. 

In fact, a $p$-value of 0.05 means there is a 1 in 20 chance that the result might have occurred my chance --- so it makes sense that if you test 20 different hypotheses, you'll likely get one statistically significant result even if none of the hypotheses you're testing are true!

This cartoon I've showed previously makes this point:

<img src="significant.png">

**$p$-hacking is when you test so many hypotheses that you become _very_ likely to get a statistically significant $p$-value for a null hypothesis that is true** --- i.e., to reach a $p$-value that incorrectly leads you to reject a true null hypothesis (e.g., thinking a treatment had an effect that actually has no effect).

Note that $p$-hacking is a _particularly_ big risk when examining heterogenous treatment effects, because in many datasets there are a ton of different subgroups you can slice and dice your dataset into. It can also come up when looking at different sets of experiments, like comparing experiments conducted in primaries vs. general elections after a meta-analysis, to see where the treatment appeared more effective.

This can result in a situation where researchers think or convince others than, even if a treatment doesn't work overall, if you look at a specific group it does. Whenever you hear a claim like this, you should worry about $p$-hacking.

The _New York Times_ did a nice story called "<a href="https://www.nytimes.com/2017/11/28/magazine/a-failure-to-heal.html?smid=pl-share" target="_blank">A Failure to Heal</a>" about how this happens in medicine, especially drug development. When a drug in testing fails to have an effect...

> a second instinct takes over: Why not try to find the people for whom the drug did work? ...
>
> This kind of search-and-rescue mission is called “post hoc” analysis. It’s exhilarating — and dangerous. On one hand, it promises the possibility of resuscitating the medicine: Find the right group of responsive patients within the trial group — men above 60, say, or postmenopausal women — and you can, perhaps, pull the drug out of the rubble of the failed study.
>
> But it’s also a treacherous seduction. The reasoning is fatally circular — a just-so story. You go hunting for groups of patients that happened to respond — and then you turn around and claim that the drug “worked” on, um, those very patients that you found. (It’s quite different if the subgroups are defined before the trial. There’s still the statistical danger of overparsing the groups, but the reasoning is fundamentally less circular.) It would be as if Sacks, having found that the three long-term responders to L-dopa happened to be 80-year-old women from one nursing home, then published a study claiming that the drug “worked” on Brooklyn octogenarians.
>
> Perhaps the most stinging reminder of these pitfalls comes from a timeless paper published by the statistician Richard Peto. In 1988, Peto and colleagues had finished an enormous randomized trial on 17,000 patients that proved the benefit of aspirin after a heart attack. The Lancet agreed to publish the data, but with a catch: The editors wanted to determine which patients had benefited the most. Older or younger subjects? Men or women?
>
> Peto, a statistical rigorist, refused — such analyses would inevitably lead to artifactual conclusions — but the editors persisted, declining to advance the paper otherwise. Peto sent the paper back, but with a prank buried inside. The clinical subgroups were there, as requested — but he had inserted an additional one: “The patients were subdivided into 12 ... groups according to their medieval astrological birth signs.” When the tongue-in-cheek zodiac subgroups were analyzed, Geminis and Libras were found to have no benefit from aspirin, but the drug “produced halving of risk if you were born under Capricorn.” Peto now insisted that the “astrological subgroups” also be included in the paper — in part to serve as a moral lesson for posterity. I’ve often thought of Peto’s paper as required reading for every medical student.

Beware of this instinct! It can be a good idea to check if a treatment has different effects for different groups, but $p < 0.05$ is a lot less meaningful when you've tested many hypotheses.

#### The value of replication

So what should we do?  It's clearly valuable to see if an effect generalizes across many studies---and it's also useful to know if certain contexts (like primaries vs. general elections) make a treatment like electoral persuasion more impactful. And we really most want to know *why* a treatment has an effect.

This is where the value of intentional **replication** of experiments can arise, in two ways:

1. Suppose a big meta-analysis of many experiments (or a sub-group analysis in a particular experiment) suggests that a particular treatment is most effective for a particular type of subject, or in a particular context).  But we are worried about *p*-hacking.


   In this case, we could *replicate* the experiment for this type of subject and context, pre-specifying our expectations about the effect we should see if the findings are real---and seeing what happens.
   
3. This type of replication is valuable for another reason: it may allow us to embed variation in the treatments that can help us assess *why* a treatment works.


   For example, as suggested above, we could build in the presence or absence of party labels in the experiment to test if this could be one reason persuasion treatments appear more effective in primaries than general elections.

    We actually talked about building in this kind of variation early in the class -- in the social pressure experiment in Week 5, we talked about how the variation in the treatments (the "Civic Duty" vs. "Hawthorne effect" vs "Self only" vs "Social Pressure/Neighbors" treatments) could help us understand *why* the treatment had an effect.

So those are two big reasons for replicating experiments---to try to make sure any differences (treatment effects) we see are real, rather than statistical artifacts, and to try to extend and revise experiments to help us understand why effects may arise.  

These strategies can help us understand the type of treatments that work, for what types of subjects and in what types of contexts---and than can help us with generalizability of our findings!
