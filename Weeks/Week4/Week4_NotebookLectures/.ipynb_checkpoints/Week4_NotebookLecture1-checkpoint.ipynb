### Part 1: Importing the Data

```r
# Read in the data.
data <- read.csv('ps3_wellness_with_POs.csv', stringsAsFactors = F)
data # Let's take a look at the data.
```

This dataset contains real data on participants in a wellness program... and a couple fake variables we've added in: the potential outcomes.

Each row in this dataset represents a unique person, and the data measures medical care costs before and after the wellness program started. Here is more information about the variables:

- `name`: Respondent Name
- `participate`: `0` if the person did not participate in the workplace wellness program; `1` if the person did participate in the workplace wellness program. (Here, take this variable to be **randomly assigned**; in Notebook Lecture 2, though, we'll distinguish self-selected actual participation from randomly assigned eligibility to participate).
- `baseline`: Amount of monthly average medical costs at baseline; that is, before the program started.
- `po_control`: Monthly cost of medical care for this person after the workplace wellness program started, measured in dollars, **if the person _did not_ participate in the program**. In this study, this is the control potential outcome (because this is what would have happened if these people did not receive the treatment; i.e., if they did not participate in the wellness program).
- `po_treat`: Monthly cost of medical care for this person after the workplace wellness program started, measured in dollars, **if the person _did_ participate in the program**. In this study, this is the treatment potential outcome (because this is what would have happened if these people received the treatment, which is the wellness program).

-----

### Analyzing Causal Effects

Let's study the effect of participation in the wellness program on the monthly cost of medical care.

First, let's calculate the unit causal effect---i.e., the treatment effect for each individual. Usually, we can't directly observe the unit causal effect, but in this data set---because we've made up the potential outcomes under treatment and potential outcomes under control for every unit---we can calculate it. 

A note on terminology: features such as the unit causal effect that we can't directly observe are often called **parameters**.  Later in the course, we will contrast parameters with **statistics** that we can only observe for some portion of the units (sometimes, random samples).

The unit causal effect is the difference between the treatment potential outcome minus the control potential outcome for a given unit. Here, a "unit" is a given person in the dataset---i.e., Alan, Ellen, Will, and so on.



```r
# Let's calculate the unit causal effect for every person and
# add the effects to the dataset as a column called "treatment_effect"
data$treatment_effect <- data$po_treat - data$po_control
data
```

Remember, in a real study, we are not going to be able to "see" the potential outcome under treatment **and** the potential outcome under control for any person.  

After all, if a given person receives the treatment (here, participates in the wellness program), we see the potential outcome under treatment. And if the person does not receive the treatment (does not participate in the wellness program), we see the potential outcome under control.  

Here, because we made up the potential outcomes, it is as if we can see both potential outcomes for every person (or "subject") in the study. 

As you can see, the treatment effect is zero for most people, but some people have non-zero treatment effects.

Let's look at the data for one person, Will, whose treatment effect isn't zero:

```r
subset(data, name == 'Will') 
```

Will, in fact, will spend **more** on medical care if he participates in the wellness program than if he does not.  

Maybe Will needs to be careful about how much he exercises!



Now let's think about everyone in this dataset, not just Will. What is the *average treatment effect* of the workplace wellness program for the people in the `data` dataset? (This isn't a trick question: what is the average or *mean* of the `treatment_effect` variable?)


```r
average.treatment.effect <- mean(data$treatment_effect) 
average.treatment.effect 
```

So on average, participation in the program increases medical costs by $0.75, or 75 cents. 

Notice however that for most people, there is no effect at all. And for some people, the effect is negative (i.e., program participation decreases medical costs). 

But Will, for whom participation increases monthly costs by $70, raises the average effect above zero. 

Lesson: the average effect is a different concept from the unit effect.

In a real data set, we won't see either the average effect for all units, or the unit effect for any single unit.  

Among people who participate in the program (i.e., for whom <code>participate</code>=1), we only observe <code>po_treat</code>, not <code>po_control</code>. 

And among people who do not participate in the program (i.e., for whom <code>participate</code>=0), we only observe <code>po_control</code>, not <code>po_treat</code>. 

In the recorded lecture (both the pdf slides and the Notebook Lecture), we will learn more about how we might nonetheless use the data we do observe to **estimate** the average causal effect---i.e., to form our best guess of what it might be.  We will also discuss reasons why our best guess might be a good guess or tend to be a bad guess. 

Note, however, that in the potential outcomes framework, we cannot use data to estimate the unit causal effect for any individual.

## Summarizing what we've learned

In this Notebook Lecture, we loaded a dataset based on a real experimental study of the effectiveness of a wellness program.  

But in order to illustrate how the potential outcomes framework works, we made up potential outcomes for each participant in the dataset. 

That is, we made one column (<code>po_control</code>) showing the "potential outcome under control" (i.e., the outcome a subject would experience if assigned to the control group). And we made another column (<code>po_treatment </code>) showing the "potential outcome under treatment" (i.e., the outcome a subject would experience if assigned to the treatment group).

Finally, we calculated the difference between each potential outcome for each subject, as shown in the column (<code>treatment_effect </code>).  This difference is the **unit causal effect** for each subject.

Of course, in real data, we won't be able to see the unit causal effect (or the average of the unit causal effects, that is, the **average causal effect**).  For each subject, we just see the outcome corresponding to the treatment condition to which each subject is assigned (i.e., according to <code>treat==1</code> or <code>treat==0</code>).
