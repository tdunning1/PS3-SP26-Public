Confidence intervals let us think about this more precisely.

### Confidence Intervals

This is why we need confidence intervals. Confidence intervals give us a range of numbers that will contain the true average treatment effect a set percentage of the time. They consist of *two* numbers: a top and a bottom. The confidence interval is the range between the top and bottom of the range.

#### 95% Confidence Intervals

By far the most common confidence intervals in social science are *95% confidence intervals*. **A 95% confidence interval has a 95% chance of including the true average treatment effect.** (Even more precisely: in 95% of experiments, the confidence interval will contain the true average treatment effect.)

Conventionally in social science, we think of a 95% confidence interval as containing **the range of _plausible_ estimates of the average treatment effect**. For example, if Sarah is really pessimistic about a treatment, she can look at the 95% confidence interval and say, "well, I think the right estimate is the very bottom of the range."

#### 95% Confidence Intervals as Containing Most Optimistic and Most Pessimistic Plausible Values

More generally, we can think of the top/bottom of a confidence interval as containing the most optimistic/pessimistic scenario that is still plausible given the data.

For example, if you're looking at the effect of a treatment on something you want to see go up (e.g., effect of mail on voter turnout), the most optimistic plausible estimate would be the top of the confidence interval. So if the confidence interval for a mailer went from a 3.2pp effect on turnout to a 6.5pp effect on turnout, 6.5pp would be the most optimistic -- or "best case" -- plausible scenario. Likewise, 3.2pp would be the most pessimistic -- or "worst case" -- plausible scenario.

#### Formula for 95% Confidence Intervals

The confidence interval is calculated from the estimate and the standard error. **Here's where the normal curve comes in.** We take the estimate in the sample as our best guess of the true target in the population; and we reason that if the sampling distribution were centered there, 95% of samples would result in estimates within 1.96 standard errors of the best guess.  

That's in turn because the sampling distribution is approximately normally distributed, at least in moderately large studies.  (Remember the central limit theorems and the examples we discussed in the pdf slides!).

So, here's how we calculate the confidence interval:
- $\text{Top of 95% Confidence Interval} = \text{Estimate} + 1.96 * \text{Standard Error}$
- $\text{Bottom of 95% Confidence Interval} = \text{Estimate} - 1.96 * \text{Standard Error}$

This means:
- The *middle* of the confidence interval is the estimate.
- The *width* of the confidence interval is determined by the standard error. For example, the lower the standard error, the *narrower* the confidence interval will be.

(Sometimes for convenience, people use 2 instead of 1.96 to calculate the confidence interval rapidly).

#### Warning: larger standard errors mean wider confidence intervals!

### Another example: drivers' licenses in India

The use of experiments (or Randomized Controlled Trials---RCTs) has exploded in research on development.  The 2019 Nobel Prize in economics was awarded jointly to Abhijit Banerjee, Esther Duflo and Michael Kremer "for their experimental approach to alleviating global poverty"

Experiments have been used to try to answer some elusive questions. Consider the difficulty of understanding factors that influence
corruption. 
    - Does payment of bribes facilitate access to basic services—and if so, how?
    - Does corruption in the delivery of services “distort” allocations—i.e., is corruption anything more than a transfer of
income from citizens to bureaucrats?

Bertrand et al. (2007) experimentally studied applications for driver’s licenses in India (see the <a href="https://academic.oup.com/qje/article-abstract/122/4/1639/1850513?redirectedFrom=fulltext">supplementary reading</a> for this week). 

Applicants for drivers’ licenses are assigned to one of three groups: (1) bonus group; (2) lessons group; or (3) control
group.  The first group was offered a bonus for obtaining a license quickly; the second was offered free driving lessons; and no offer of help was made to the third, control group. 

Both the bonus and lesson groups were more likely to obtain licenses. However, bonus group members were more likely to make extralegal payments and to obtain licenses without knowing how to drive. Extralegal payments happen through private intermediaries (“agents”), who the researchers showed can circumvent procedures such as the driving test.

Here's what the researchers found in more detail:


<img src="DriversLicenseInIndia_1.png" />

We can see, for example, that according to the point estimates, those assigned to the "bonus" group were 25 percentage points more likely to obtain a license, 42 percentage points more likely to do so in 32 days or less, 13 percentages points more likely to obtain the license without taking an exam, and 29 percentage points more likely to do so without having anyone teach them to drive.  (The researchrs also find evidence elsewhere that this is because applicants are paying "agents" -- intermediaries -- to pay bribes on their behalf).

But what range of estimates is consistent with chance variation?  Take the results highlighted in Column 2.  The point estimate is .25 (or 25 percentage points), while the standard error (in parentheses) is 0.05 (or 5 percentage points). 

Using plus or minus 2 SEs rather than 1.96, for convenience, gives us an approximate 95% confidence interval of 25 percentage points, plus or minus 10 percentage points.  

The confidence interval is very far from zero percentage points. The interval's minimum value is 15 percentage points.  That's still a pretty big effect. So a skeptic doesn't have much scope to argue, at least not on the gounds that the apparent effect is only due to chance error.

### Another example: an experiment on electing women

Finally, let's examine confidence intervals using a dataset from a neat experiment done in Utah on how to elect more women to political positions. 

Women are underrepresented in political positions nationwide, and there is a lot of research on how to involve more women in politics---especially how to encourage women to run for political office.

The experiment we will analyze here describes one such effort. The experiment was conducted by professors at Brigham Young University in partnership with the Utah Republican Party in 2014. The Utah Republican Party wanted more women to become active in the Party.

To accomplish this, they embedded an experiment in party precinct conventions. (A precinct is a small local area, corresponding to the area containing all the people who vote at a particular polling place.) At party precinct conventions, they select who will go on to represent the precinct at the state's Republican convention. This position is called being a *state delegate*.

Each precinct has a precinct chair who runs the party's precinct convention in that precinct. In the data, over 2,000 precinct chairs were randomly assigned to receive one of four letters from the state party chair prior to these neighborhood caucus meetings. The Republican Party of Utah sent the precinct chairs a letter prior to the convention, but the contents varied by treatment group:

- Control group: letter did not mention women and running for state delegate
- Supply group: letter instructed precinct chair to recruit 2-3 women to run as state delegate
- Demand group: letter instructed precinct chair to read a message from the party at the precinct meeting asking women to run
- Both group (Supply + Demand): both of the above

Let's take a look at the data:

```r
library(estimatr)
data <- read.csv('ps3_week7_electing_women.csv')
head(data)
```

Here is a quick rundown of what each column means:

- `unique_id`: Precinct ID
- `treat`: treatment variable
    - `'control'`: control group
    - `'supply'`: supply group; party chair instructed to recruit 2-3 women
    - `'demand'`: demand group; party chair reads letter at precinct convention
    - `'both'`: a fourth group getting both the supply and demand treatments; party chair instructed to read letter *and* to recruit 2-3 women
- `prop_sd_fem2014`: **Outcome**: Proportion of 2014 elected state delegates from that precinct who were women
- `sd_onefem2014`: 1 if at least one woman was selected; 0 otherwise
- `county` : County name in Utah
- `pc_male`: 1 if precinct chair is male; 0 otherwise (precinct chair is person who runs precinct meeting, would read letter if assigned to do so, etc.)

```r
# And here's the values of the treatment variable again:
table(data$treat)
```

### Computing a confidence interval

Now let's run the `difference_in_means()` function.

```r
demand.vs.control.dim <- difference_in_means(prop_sd_fem2014 ~ treat, data,
                                             condition1 = 'control', condition2 = 'demand')
demand.vs.control.dim
```

`difference_in_means()` prints the 95% confidence interval for us! (It doesn't say it is a 95% confidence interval, but it is; 95% confidence intervals are the norm, so we assume a confidence interval is a 95% confidence interval unless stated otherwise.)

What does this confidence interval mean?

- **Estimate**: Our best guess is that just reading this letter increased the proportion of women elected at a precinct convention by around 1.9 percentage points.
- **CI Lower**: The lowest value that is plausible is a 2.8 percentage point *decrease*; it is plausible that the letter actually decreased the number of women elected.
- **CI Upper**: The highest value that is plausible is a 6.6 percentage point increase.

Suppose someone was a really big optimist about the power of reading these letters. The most optimsitic they can plausibly be after doing the experiment is that these letters increase the number of women by around 6.6 percentage points.

#### Retrieving the ends of the 95% confidence interval from `difference_in_means()`

We can actually save the ends of the 95% confidence interval as numbers, like this, where `demand.vs.control.dim` is the variable where we stored the output of `difference_in_means()` above (this could be any name):

```r
demand.vs.control.95ci.high.end <- demand.vs.control.dim$conf.high
demand.vs.control.95ci.high.end
```

```r
demand.vs.control.95ci.low.end <- demand.vs.control.dim$conf.low
demand.vs.control.95ci.low.end
```

### Using confidence intervals

Confidence intervals are a good way to translate a range of uncertainty into numbers that everyday people can understand and make decisions using:

Confidence intervals can be used to answer questions like:

- How many units need to be treated to change one outcome?
- What is the cost per changed outcome?
- How many units would change outcomes if a certain number are treated?

Here's how to do one of these as an example:

#### Example: Expected number of units that would change outcomes if a certain number of units are treated

Suppose we were to roll out the "demand" treatment to 500 precincts? How many new women would we expect to be elected?

$\text{Expected Units With Different Outcomes} = \text{Number Units Treated} * \text{Expected Effect Per Unit}$

```r
at.least.one.woman <- difference_in_means(sd_onefem2014 ~ treat, data,
                                          condition1 = 'control', condition2 = 'demand')
at.least.one.woman
```

Remember we can access the upper and lower end of a confidence interval like this:

```r
at.least.one.woman$conf.low
at.least.one.woman$conf.high
```

So, translating those estimates, if 500 precincts received this treatment, we'd expect to see the number of _additional_ precincts that elect at least one woman to be in this range, from least to most optimistic:

```r
500 * at.least.one.woman$conf.low # most pessimistic
500 * at.least.one.woman$conf.high # most optimistic
```

Using this information, a decision-maker might have an easier time deciding if this is worth it...
