## Part 1: The effects of co-ethnicity and cousinage

For Part 1 of this activity, we will be revisiting the `difference_in_means()` function from the `estimatr` package in order for us to make some inferences about the data from Dunning and Harrison's (2010) study.

Recall that Dunning and Harrison study whether the presence of a cross-cutting cleavage -- that is due to the fact that people have "joking cousins" who are members of different ethnic groups -- helps reduce ethnic voting in politics.

In this study, participants in an experiment in Bamako watched a video of person described as a potential political candidate.  Participants watched the same video but were told different things about the candidate's last name (i.e. surname or "patronym").  What they were told depended on their own last name, as well as the treatment condition to which they were assigned.  

Because last name conveys information about both ethnic identity and cousinage ties in Mali, this experimental procedure assigned participants to one of the following treatment conditions at random:
  1. Same ethnicity, joking cousin
  2. Same ethnicity, not joking cousin
  3. Different ethnicity, joking cousin
  4. Different ethnicity, not joking cousin

(There are two other treatment conditions mentioned in the paper, but we won't analyze those here).

Now, let's take a look at the data.  

**Note: we are going to use a slightly different version of the data than in Notebook Lecture 1.**

In particular, **here we will have two variables that together record the treatment conditions:`treat_assign_coethnic` and `treat_assign_joking_cousin`. Since each of them can take two values (`1` and `0`), they represent the four treatment conditions above**.

```r
#RUN THIS CELL
library(testthat)
library(estimatr)

data_dh <- read.csv('DunningHarrison_str.csv')
head(data_dh)
```

Here are what the variables mean.  This is similar to the dataset in Notebook Lecture 1 with the exception of `treat_assign_coethnic` and `treat_assign_joking_cousin`

- `participantid`: a unique identifier for each subject (each person in the experiment)
- `vote_prefer`: 'On a scale from 1 to 7, how much does this speech make you want to vote for (name of politician)?' 7 = most likely to want to vote for; 1 = least likely to want to vote for
- `treat_assign_coethnic`: `1` if the subject (meaning person in the experiment) was randomly assigned to be shown a candidate with a co-ethnic surname, `0` otherwise
- `treat_assign_joking_cousin`: `1` if the subject was randomly assigned to a treatment group with a candidate who was a joking cousin, `0` otherwise
- `actor`: this number indicates which actor each participant watched. Speeches were filmed in July 2008 in NYC, with two Malians actors. (`1`: Mamadou Doumbia; `2`: Bamadou Diallo)
- `interviewer`: a numeric indicator for the particuar enumerator (interviewer) who collected the data
- `female`: equals `1` if the respondent identifies as female, `0` otherwise
- `education`: `1` if the subject received higher education, `0` otherwise
- `vecu_ailleurs`: `1` if subject has lived elsewhere from Bamako, `0` otherwise
- `inscrite`: `1` if subject is registered to vote, `0` otherwise
- `cousin_subject`: `1` if subject **believes** the actor is their joking cousin after viewing the speech, `0` otherwise

We won't use all of the variables in this analysis, but this gives you a sense of some of the data in the dataset.

**Again: the treatment condition is represented by two variables here, `treat_assign_coethnic` and `treat_assign_joking_cousin`. Since each of them can take two values (`1` and `0`), they represent the four treatment conditions**.

## PART 2: Comparing the effects of co-ethnicity and cousinage

So far, we've analyzed evidence on (1) the effect of the co-ethnic condition and (2) the effect of the joking cousin condition.  But what if we want to know which of these treatment conditions has a stronger influence on voting preferences?

To help us answer this question, we might want to compare the voting preferences of participants assigned to the "same ethnicity, not joking cousin" treatment (that is, treatment 2 in the list of treatments at the beginning of this activity) to those assigned to the "different ethnicity, joking cousin" treatment (that is, treatment 3 in the list at the beginning of this activity).

This comparison might help us answer questions like the following: Are people more prone to vote for a cousin who is from a different ethnic group, or a co-ethnic who is not their cousin?  Or are they equally likely to vote for either type of candidate, on average?

To do this kind of analysis, we need first to do some data manipulation. We are going to create a subset only of those participants assigned to one of these two treatment conditions (i.e., to treatment condition 2 or 3 listed at the top of the Notebook).  

The following code creates this subset.  The code uses one piece of syntax we haven't learned yet, so the code chunk does this for you.  

But here is a brief explanation. Remember that you learned how to subset on two conditions simultaneously using `&` -- the "AND" operator -- inside the `subset()` function. 

The code below uses an additional operator `|` -- the "OR" operator -- to subset on participants who meet either one OR another condition -- in particular, the "same ethnicity, not joking cousin" treatment OR the "different ethnicity, joking cousin" treatment

In words, the code creates a subset of the dataset `data_dh` containing participants who were
- Assigned to treat_assign_coethnic==1 AND treat_assign_joking_cousin==0

    OR

- Assigned to treat_assign_joking_cousin==0&treat_assign_joking_cousin==1.

Thus, the subset only has participants assigned to treatment conditions 2 and 3 in the list above.

(Note: you don't need to use the OR operator, `|`, in this activity yourself, other than by running the code chunk below.  But we'll use it in the future)

```r
# Run this code chunk

data_dh_subset <- subset(data_dh, (treat_assign_coethnic==1&treat_assign_joking_cousin==0|treat_assign_coethnic==0&treat_assign_joking_cousin==1))
```

By the way, you can check this worked with `table()` using the original dataset and the subsetted dataset.  Here's the distribution of the treatment assignments in the original dataset:

```r
table(data_dh$treat_assign_coethnic,data_dh$treat_assign_joking_cousin)
```

And here's the distribution in the subset:

```r
table(data_dh_subset$treat_assign_coethnic,data_dh_subset$treat_assign_joking_cousin)
```

Notice that participants in the off-diagonal cells of the first table disappeared -- as they should have, because those are people who were assigned either to a non-coethnic non-cousin candidate (top left of the first table) or to a co-ethnic cousin candidate (bottom right of the first table).
