## Week 8, Group Activity 1: Covariate Balance Tests

In this activity, we'll practice using **covariate balance tests** to assess a core assumption of natural experiments: **as-if random** treatment assignment.

As-if random is defined as the **independence of treatment assignment and potential outcomes**.



### As example: land titles in Argentina

We're going to continue working with the <a href="https://www.jstor.org/stable/25098841">Di Tella et al.</a> study of the impact of land titles in Argentina, as we did in Notebook Lecture 1. 

However, here we are going to start with a dataset that records characteristics of the parcels that squatters occupied, rather than squatters themselves.  

In particular, we will start by replicating Table III in the paper below.  As a reminder, here's what the table in their paper looked like:


<img src="DiTellaBalance2.jpg"/>

In this table, Di Tella et al. assess the balance of four pre-treatment covariates, comparing parcels where the property right offer was extended (property right availability=1) to those where it was not (property right availability=0). The pre-treatment covariates are: distance to a creek (in blocks); distance to nonsquatted area (in blocks); parcel surface (in square meters); and whether the block is on a corner.  

The idea is to assess whether there might be some confounding, if absentee owners of the plots decided---contra as-if random assignment---to protest expropriation of their land in court only if their parcels are particularly attractive or valuable.  If that were true, it could introduce confounding since the quality of the parcel might, in addition, be related to some of the outcome variables in the study.
