### How do we validate a "natural experiment?"

In this Notebook Lecture, we'll learn about one widely used technique that researchers use to try to assess the claim that a study is a "natural experiment."  

This is the technique of **covariate balance tests.**

Let's talk about why researchers do these tests.

Remember, natural experiments are **observational studies** (those in which there is no researcher-induced intervention). 

However, unlike many observational studies, in natural experiments the assignment to treatment is either actually **randomized**, e.g., with a lottery---or it is alleged to be merely **"as-if random."** 

Either way, because the researcher did not control the randomization, we want to check as best we can whether as-if random actually holds.  After all, if it does, we can analyze the natural experiment just as if it were a real experiment---using the simple and transparent techniques we have learned so far in PS 3.  

For example, we might want to take a simple difference of means across the treatment and control groups.  Even if we use more complicated methods, as-if random would mean we don't have to worry about confounding.

As we discussed in the in-class (pdf) lecture slides, as-if random assignment means the following: **treatment is assigned independently of potential outcomes.**  

Note that this condition is implied by using a randomized lottery to determine treatment assignment: if you flip a coin to decide who goes to the treatment group, there should be any association between potential outcomes and treatment assignment.  

In the Hajj study we discussed in the in-class (pdf) lecture slides, for example, people who are already tolerant should be as likely to lose in the visa lottery as win. 

As-if random might hold in some other observational studies, too, even if there is no true randomization.  We've already talked about some examples.

If true, as-if random gets rid of the problem of selection bias---because the potential outcomes of those assigned to treatment shouldn't be systematically different from the potential outcomes of those assigned to control, on average.  

Then, a simple difference of means should be an unbiased estimator for the average treatment effect. Just like in a real randomized experiment!

Now, if as-if random holds, we should **expect** average potential outcomes to be about the same in the treatment and the control groups. 

Of course, they may differ in any given experiment due just to chance error---but since we now know how to quantify uncertainty, that is not a problem!

### Why use covariates at all?  Why not just test for the balance of potential outcomes?

The problem, however, is that we do not get to see all the potential outcomes!  This is what we've called the **fundamental problem of causal inference.**  

Thus, in the control group, we don't see potential outcomes under treatment.  And in the treatment group, we don't see potential outcomes under control.

Think back to our box model for experiments.  There's a reason we slashed out the potential outcomes under control in the treatment group sample, and we slashed out the potential outcomes under treatment in the control group sample.

That means **we can't directly compare the average potential outcomes under control in the treatment and control groups.**  Same goes for the average potential outcomes under treatment.

This is why researchers use **covariates** instead.  Covariates are variables you can think of as being fixed before the (natural) experiment took place.  They should not themselves be influenced by the treatment (they are not potential outcome variables).

The hope is that if *covariates* are balanced statistically across the treatment and control groups, potential outcomes may be as well.  This is more likely to be true, however, if in your tests you use covariates that might be closely associated with potential outcomes!  We'll talk more about this later in this Notebook Lecture.

### As example: land titles in Argentina

Let's use the example of land titles in Argentina, i.e., the study of <a href="https://www.jstor.org/stable/25098841">Di Tella, Galiani and Schargrodsky<a> (also on bCourses) that we discussed in class and that Prof. Dunning describes in Chapter 1 of *Natural Experiments in the Social Sciences: A Design-Based Approach* (which is part of the reading for this week).

In the paper, Di Tella et al. assess the balance of eight pre-treatment covariates, using data they gathered on the age, gender, education and other variables **of the original squatter at the time property titles were assigned**.

Remember, here's what the table in their paper looked like:


<img src="DiTellaBalance1_v2.jpeg"/>

Here, Di Tella et al.  show the means and the standard errors in the treatment and control group samples of eight pre-treatment **covariates**: these variables measure the age, gender, nationality, and education of the original squatter, as well as the nationality and education of the squatter's father and the nationality and education of the squatter's mother. The treatment and control groups are defined by "property right availability" -- whether a squatter was or was not offered a property title.

Next, we are going to replicate some parts of this table so that we can interpret it.  Let's look at the data (this is a selection of the columns in the replication data set for the paper).

```r
library(estimatr)
DiTella_squatters <- read.csv('DiTella_squatters.csv')
head(DiTella_squatters)
```

(Note that you can scroll over to the right to see all the variables).

Here is a quick rundown of what each column means.  In the first three columns, we have two ID variables and the treatment variable.

- `parcelId`: an ID variable for the parcel on which the squatter lived.
- `householdId`: an ID variable for the squatter.
- `propertyOffer`: whether the squatter was or was not offered a property title (based on whether the government's offer of a property title was or was not challenged in court)
- `propertyRight`: whether the squatter actually obtained a title to their property

Next, we have the covariates:
- `ageOrigSquatter`: the age of the squatter, in years
- `femaleOrigSquatter`: the gender of the squatter (female=1, male=0)
- `argentineOrigSquatter`: whether the original squatter was Argentine (1=yes, 0=no) (In this part of metropolitan Buenos Aires, there were many immigrants from Bolivia and elsewhere in Latin America)
- `educationYearsOrigSquatter`: the years of education of the original squatter
- `argentineOrigSquatterFather`: whether the original squatter's father was Argentine
- `educYearsORigSquatterFather`: years of education of the original squatters father
- `argentineOrigSquatteMother`: whether the original squatter's mother was Argentine
- `educYearsORigSquatterMother`: years of education of the original squatters mother

Notice that all the covariates can be thought of as "fixed" -- their values are thought of as being already determined at the time the squatting took place.

Finally, we have five outcome variables.  We won't look at these in this lecture, but you will do so in the in-class activity this week.
- `successAlone`: Takes the value 1 if the answer to the question "Do you believe that it is possible to be successful on your own or a large group that supports each other is necessary?" was "It is possible to be successful on your own"; and 0 if the answer was "A large group is necessary to be successful." 
- `moneyImportant`: Takes value 1 if the answer to the question "Do you believe that having money is important to be happy?" was either "Indispensable to be happy," "Very important to be happy" or "Important to be happy"; and 0 if the answer was "Not important to be happy." 
- `effortBetter`: Takes the value 1 if the answer to the question "In general, people who put effort working end up much better, better, worst or much worst than those who do not put an effort?" was "Much better than those that do not put an effort" or "Better than those that do not put an effort"; and 0 if the answer was "Worst than those that do not put an effort" or "Much worst than those that do not put an effort." 
- `trustOthers`: Takes the value 1 if the answer to the question "In general, in our country, would you say that one can trust other people or that people cannot be trusted?" was "You can trust others" and 0 if the answer was "You cannot trust others." 
- `marketBeliefs`: An index that is the individual's sum (ranging from 0 to 4) of the four previous outcome variables.


```r
# Here are the values of the treatment variable:
table(DiTella_squatters$propertyOffer)
```

### Replicating Table II

Let's replicate just the averages in the first row (for the "age" variable) using the "old" method -- that will allow us to see just the means in each group, and their difference.


```r
# The average of the age variable in the control group

# First subset to the control group
control <- subset(DiTella_squatters, propertyOffer==0)

# Now calculate the average
# Note that na.rm tells R to ignore missing values in calculating the average
ave_age_control <- mean(control$ageOrigSquatter, na.rm = TRUE)

ave_age_control # print the result
```

```r
# Now subset the treatment group

treatment <- subset(DiTella_squatters, propertyOffer==1)

# Calculate the average of the age variable
# Again, na.rm tells R to ignore missing values in calculating the average

ave_age_treatment <- mean(treatment$ageOrigSquatter, na.rm = TRUE)

ave_age_treatment
```

```r
# Now calculate the difference of means

ave_age_treatment - ave_age_control
```

Note that this difference is the same as that given in the final column of the first row of Table II---except that for some reason, the authors calculate the difference as control minus treatment, rather than treatment minus control!  So the sign is different.  The number is the same, however.  You should calculate the differences as treatment minus control, as this is standard.


Now let's run the `difference_in_means()` function on the age variable to get the estimate and standard error in the first row of the final column of Table II.

```r
age.dim <- difference_in_means(ageOrigSquatter ~ propertyOffer, DiTella_squatters)
age.dim
```

The difference is just what we found above (and is the same as in Table II, though again the sign is different since in Table II, they subtract the treatment group mean from the control group mean...)

As for the standard error, it is 1.208, just like in the table (standard errors are always positive -- they are the square root of the variance, which is a squared quantity and thus positive).

### What does this mean?

Using the output from the difference of means function, we can now conduct a hypothesis test.  You can state the null hypothesis as something like "the treatment and control groups are drawn from the same population" or "treatment is assigned independently of covariates" though this isn't exactly how we've set up the null hypothesis before, when testing for average treatment effects. \

First, notice that the t-value is less than 1.96 in absolute value: it is about 1.268.  We could calculate this directly, too:

```r
t_stat <- age.dim$coefficients/age.dim$std.error
t_stat # print the t-stat
```

Thus, the *p*-value is greater than 0.05 -- it is about 0.206:

```r
age.dim$p.value
```

So, what does this mean?  It means we **fail to reject** the null hypothesis.  

If we instead reject the null hypothesis, it would mean the following: if as-if random holds, it would be unusual to see a difference of means on the pre-treatment covariate that is as large as the one we actually see.  

That's not what we find here. Instead, here we "fail to reject" the null hypothesis. 

This, of course, does not mean that the null hypothesis is true!  But we cannot reject it.  

The logic is a little different than when we are testing for treatment effects.  There, if we believe the alternative hypothesis, we would want to reject the null (so we can argue against our skeptic Sarah). 

Here, we are instead usually "hoping" **not** to reject the null hypothesis.  Failing to reject doesn't prove that as-if random holds----but it is useful evidence.  

Conversely, if we could reject the null hypothesis, it would make as-if random less plausible.  More on this below.

We can finish the replication of Table II by running the following code.  I don't expect you to learn all this code but do want you to understand what it is doing.  

First, we are going to store the names of all the covariates included in Table II. 

```r
# Store the names of the covariates in the DiTella_squatters data in an object called cov_names

suppressWarnings(library(dplyr)) # load the dplyr package without any warnings from R

cov_names <- DiTella_squatters %>%
  dplyr::select(ageOrigSquatter, femaleOrigSquatter, 
                argentineOrigSquatter, 
                educationYearsOrigSquatter,
                argentineOrigSquatterFather,
                educYearsOrigSquatterFather,
                argentineOrigSquatterMother,
                educYearsOrigSquatterMother) %>% names
```

```r
# Now we run a function that calculates the difference of means for each of the covariates
# and stores them in an object called bal_tab3

bal_tab3 <- sapply(cov_names, function(cov){
  form <- as.formula(paste(cov, "~ propertyOffer"))
  fit <- difference_in_means(form, data = DiTella_squatters)
  c(estimate = fit$coefficients, se = fit$std.error)
}) %>% t() %>% as.data.frame()
```

```r
# Write a table that replicates the final column of Table II in Di Tella et al.

# YOU DON'T NEED TO RUN THIS CELL -- if you do, the output immediately after the cell won't look
# very intelligible (the intelligle output is pasted below)

suppressWarnings(library(kableExtra)) # load the kableExtra package without warnings from R

results <- bal_tab3 %>%
  mutate(formatted = paste0(round(estimate.propertyOffer, 3), " (", round(se.propertyOffer, 3), ")")) %>%
  select(formatted) %>%
  kable(
    col.names = c("Estimate (SE)"),
    caption = "Balance Table: Difference in Means by Property Offer",
    booktabs = TRUE
  ) %>%
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

You can compare the estimates and standard errors to the final column of Table II above.  They are the same (again, except that the signs are reversed in Table II).

Notice also---and here is the main conclusion--that if you divide the estimates by the standard errors, the result for all the covariates is less than 1.96 in absolute value.  In other words, we do not reject as-if random for any of the eight covariate balance tests---as we would hope.

### How powerful are these tests?

It's natural to ask how dispositive the results of these tests are.  I.e., do they "prove" that treatment is assigned as-if at random?  That is, do they prove that Di Tella et al.'s study is a valid natural experiment?

No, they do not.  We need to look at many kinds of evidence to reach this conclusion---including the qualitative evidence we discussed in the in-class (pdf) lecture.  And even then, we could reach the wrong conclusion.  But that's why it's so important to test the assumption of as-if random using as many different kinds of evidence as we can come up with.

There is a further point that we should make about these kinds of balance tests.  Remember that we want to test whether treatment assignment is independent of potential outcomes.  But we are conducting the tests using not **potential outcomes** but instead **covariates**.

It is natural therefore to think that the tests are more powerful if the covariates we happen to measure are strongly associated with the actual potential outcomes.  Covariates that are strongly predictive of potential outcomes are called **prognostic**. 

And indeed, it is true that the more prognostic are the covariates we use in our balance tests, the more powerful the tests will be.  This is the point of a <a href="https://arxiv.org/pdf/2205.10478v2">recent paper</a> that co-authors and I have written (that is forthcoming in the journal *Political Analysis*). 

It turns out that you can assess empirically how prognostic a covariate is.  For example, you can assess the association between the covariate and potential outcomes under control using data from the control-group sample.

The correlation coefficient---which we learned about earlier in the semester---is a useful tool.  For example, here is the correlation between the outcome variable `successAlone` and gender:


```r
cor(control$successAlone,control$female)
```

That is a moderately big, but not overwhelming correlation. So gender is only moderately **prognostic** for this potential outcome under control for this variable.

It might also be useful to look at a two-way table between these variables.

```r
table(control$successAlone,control$female)
```

We will talk more next week about how to find the most prognostic possible covariates for balance tests. (Hint: often, the *past* or *lagged* value of an outcome variable is very predictive of its future value).

And this is an issue you should bear in mind when assessing the balance tests you will find in political science papers (or anywhere else!).
