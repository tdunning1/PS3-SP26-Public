As previewed in this week's lectures, today we will analyze data from a really interesting experiment involving over 100,000 US voters.

In this experiment, the authors sent out mailings to registered voters encouraging them to vote. Voters were randomized to either no mailing (the control group) or one of several different mailings, described below. The mailers used increasingly strong "social pressure" to encourage voters to vote.

We'll practice two techniques for analyzing this experiment -- the "old" method of calculating averages in different treatment groups directly and then comparing them; and the "new" method using using the `difference_in_means()` function from the `estimatr` package to calculate estimated effects.

Going forward, we will usually use the new method, but the old method is helpful for seeing the individual averages in each group and understanding what they estimate.
